## Basic Python Datatypes
|Data types|Meaning|Examples|
|---|---|---|
|int| Whole Numbers | 10, -5, 100|
|float|Decimal Numbers|10.3, 59.56789|
|complex|Complex numbers|
|bool|true/false|True, False|
|string|characters|"string", 'animals'|
|list|mutable list of elements|[1,2,3,"test",'^&**']|
|tuple|immutable list of elements|(1,2,3,"test",'^&**')|
|dictionary|key:value pair of elements|{1:2, 3:4}


```python
name = "Rahul" #String
age = 20 #int
marks = 85.5 #float
is_pass = True #bool
print(name)
print(age)
print(marks)
print(is_pass)
```
