**Examples:**

Linear Data Structures (Sequential / 1D array-backed layout)

|  | Data Structure | How It Organises Data | Common Use Case |
|---|---|---|---|
|  | Array | Contiguous memory blocks accessed by index. | Storing a fixed list of numbers. |
| | String | A sequential array specifically for characters. | Text processing and word search. |
| | Linked List | Nodes connected sequentially via pointers. | Building dynamic playlists. |
| | Stack | Last-In, First-Out (LIFO) sequential access. | "Undo" operations in text editors. |
| | Queue | First-In, First-Out (FIFO) sequential access. | Print job scheduling. |
| | Hash Table | Key-value pairs mapped onto a flat array. | User login verification (Username $\rightarrow$ Password). |


Non-Linear Data Structures (Hierarchical / Interconnected networks)

|  | Data Structure | How It Organises Data | Common Use Case |
|---|---|---|---|
|  | Tree | Parent-child nodes branching from a single root. | File directories on a computer. |
| | Heap | Tree-based structure optimized for min/max values. | Priority queues in operating systems. |
| | Graph | A network of nodes connected by flexible edges. | Social networks and Google Maps routes. |



# 2. Write a Program that uses function to perform the following operations on array:
1. Creation
2. Insertion
3. Deletion
4. Traversal

```python
%%writefile expone.cpp
#include <iostream>

using namespace std;

// --- Function Declarations ---
// Function to populate an array up to 'count' elements
void creation(int arr[], int count);

// Function to insert a value at a specific position
void insertion(int arr[], int& size, int pos, int val, int capacity);

// Function to traverse and print elements of an array
void traverseArray(const int arr[], int size);

// Function to delete an element by index
void deleteByIndex(int arr[], int& size, int indexToDelete);


// --- Function Definitions ---

// Function to populate an array (simplified for initial creation)
void creation(int arr[], int count) {
    for (int i = 0; i < count; i++) {
        arr[i] = (i + 1) * 10;  //Populate with multiples of 10 for clarity
    }
}

// Function to insert a node at a specific position
void insertion(int arr[], int& size, int pos, int val, int capacity) {
    if (size >= capacity) {
        cout << "Error: Array is full. Cannot insert." << endl;
        return;
    }
    if (pos < 0 || pos > size) { // pos can be 'size' to insert at the end
        cout << "Error: Invalid position for insertion." << endl;
        return;
    }

    // Shift elements to the right to make space
    for (int i = size; i > pos; --i) {
        arr[i] = arr[i - 1];
    }

    arr[pos] = val; // Insert the new value
    size++;         // Increment the logical size
}

// Function to traverse and print the array
void traverseArray(const int arr[], int size) {
    if (size == 0) {
        cout << "Array is empty." << endl;
        return;
    }
    for (int i = 0; i < size; ++i) {
        cout << arr[i] << " ";
    }
    cout << endl;
}

// Function to delete a node at a specific position
void deleteByIndex(int arr[], int& size, int indexToDelete) {
    if (size == 0) {
        cout << "Error: Array is empty. Cannot delete." << endl;
        return;
    }
    if (indexToDelete < 0 || indexToDelete >= size) {
        cout << "Error: Invalid index for deletion." << endl;
        return;
    }

    // Shift elements to the left to fill the gap
    for (int i = indexToDelete; i < size - 1; ++i) {
        arr[i] = arr[i + 1];
    }

    size--; // Decrement the logical size
}


int main() {
    const int max_capacity = 10;
    int* arr = new int[max_capacity]; // Allocate array with max capacity
    int current_size = 0; // Initialize current number of elements to 0


    // --- Creation ---
    cout << "Creating initial array with 5 elements..." << endl;
    current_size = 5; // Set initial size
    creation(arr, current_size); // Populate the first 'current_size' elements
    cout << "Array after creation: ";
    traverseArray(arr, current_size);

    // --- Insertion ---
    int insert_val = 99;
    int insert_pos = 2; // Insert at index 2
    cout << "\nInserting " << insert_val << " at position " << insert_pos << "..." << endl;
    insertion(arr, current_size, insert_pos, insert_val, max_capacity);
    cout << "Array after insertion: ";
    traverseArray(arr, current_size);

    /*
    insert_val = 77;
    insert_pos = current_size; // Insert at the end
    cout << "\nInserting " << insert_val << " at position " << insert_pos << " (end)..." << endl;
    insertion(arr, current_size, insert_pos, insert_val, max_capacity);
    cout << "Array after inserting at end: ";
    traverseArray(arr, current_size);

    // Try to insert when full (for demonstration of error handling)
    cout << "\nAttempting to fill the array and insert beyond capacity..." << endl;
    while (current_size < max_capacity) {
        insertion(arr, current_size, current_size, (current_size + 1) * 100, max_capacity);
    }
    cout << "Array after filling: ";
    traverseArray(arr, current_size);
    insertion(arr, current_size, 0, 0, max_capacity); // This should show error
    cout << "Array after attempted insertion when full: ";
    traverseArray(arr, current_size);
    */


    // --- Deletion ---
    int delete_idx = 0;
    cout << "\nDeleting element at index " << delete_idx << "..." << endl;
    deleteByIndex(arr, current_size, delete_idx);
    cout << "Array after deletion at index " << delete_idx << ": ";
    traverseArray(arr, current_size);

    /*
    delete_idx = current_size - 1; // Delete last element
    cout << "\nDeleting element at index " << delete_idx << " (last element)..." << endl;
    deleteByIndex(arr, current_size, delete_idx);
    cout << "Array after deleting last element: ";
    traverseArray(arr, current_size);

    // --- Traversal (Explicit Demonstration) ---
    cout << "\nExplicit Traversal of the current array: " << endl;
    traverseArray(arr, current_size);
    cout << "Traversal completed!" << endl;

    // Clean up dynamically allocated memory
    delete[] arr;
    arr = nullptr;
    */

    return 0;
}
```

# 3. Write a Program that uses function to perform the following operations on Linked-List:
1. Creation
2. Insertion
3. Deletion
4. Traversal

```python
%%writefile expotwo.cpp
#include <iostream>

using namespace std;

// Node structure for the linked list
struct Node {
    int data;
    Node* next;

    // Constructor
    Node(int val) : data(val), next(nullptr) {}
};

// Function to traverse and print the linked list
void traverseLinkedList(Node* head) {
    Node* current = head;
    while (current != nullptr) {
        cout << current->data << " ";
        current = current->next;
    }
    cout << endl;
}

// Function to create a linked list from an array
Node* createLinkedList(int arr[], int n) {
    if (n == 0) {
        return nullptr;
    }
    Node* head = new Node(arr[0]);
    Node* current = head;
    for (int i = 1; i < n; ++i) {
        current->next = new Node(arr[i]);
        current = current->next;
    }
    return head;
}

// Function to insert a node at a specific position
Node* insertNode(Node* head, int data, int position) {
    Node* newNode = new Node(data);
    if (position == 0) {
        newNode->next = head;
        return newNode;
    }

    Node* current = head;
    for (int i = 0; i < position - 1 && current != nullptr; ++i) {
        current = current->next;
    }

    if (current == nullptr) {
        cout << "Position out of bounds for insertion." << endl;
        delete newNode;
        return head;
    }

    newNode->next = current->next;
    current->next = newNode;
    return head;
}

// Function to delete a node at a specific position
Node* deleteNode(Node* head, int position) {
    if (head == nullptr) {
        cout << "List is empty. Cannot delete." << endl;
        return nullptr;
    }

    if (position == 0) {
        Node* temp = head;
        head = head->next;
        delete temp;
        return head;
    }

    Node* current = head;
    for (int i = 0; i < position - 1 && current != nullptr; ++i) {
        current = current->next;
    }

    if (current == nullptr || current->next == nullptr) {
        cout << "Position out of bounds for deletion." << endl;
        return head;
    }

    Node* temp = current->next;
    current->next = temp->next;
    delete temp;
    return head;
}

int main() {
    int arr[] = {10, 20, 30, 40, 50};
    int n = sizeof(arr) / sizeof(arr[0]);

    cout << "Creating Linked List..." << endl;
    Node* head = createLinkedList(arr, n);
    cout << "Original Linked List: ";
    traverseLinkedList(head);

    cout << "\nInserting 25 at position 2..." << endl;
    head = insertNode(head, 25, 2); // Insert 25 at index 2 (0-indexed)
    cout << "Linked List after insertion: ";
    traverseLinkedList(head);

    cout << "\nDeleting node at position 0..." << endl;
    head = deleteNode(head, 0); // Delete node at index 0
    cout << "Linked List after deletion: ";
    traverseLinkedList(head);

    cout << "\nDeleting node at position 3..." << endl;
    head = deleteNode(head, 3); // Delete node at index 3 (originally 4th element)
    cout << "Linked List after another deletion: ";
    traverseLinkedList(head);

    /*
    // Clean up memory (optional for this simple demo, but good practice)
    while (head != nullptr) {
        Node* temp = head;
        head = head->next;
        delete temp;
    }
    */

    return 0;
}
```

# 4. Write a Program that implements stack (its operation using array).

```python
%%writefile expthree.cpp
#include <iostream>
#include <limits> // Required for numeric_limits

using namespace std;

const int maxSize = 5; // Maximum size of the stack

class Stack {
private:
    int arr[maxSize];
    int top; // Index of the top element

public:
    Stack() {
        top = -1; // Initialize top to -1 indicating an empty stack
    }

    // Check if the stack is empty
    bool isEmpty() {
        return top == -1;
    }

    // Check if the stack is full
    bool isFull() {
        return top == maxSize - 1;
    }

    // Push an element onto the stack
    void push(int value) {
        if (isFull()) {
            cout << "Error: Stack Overflow! Cannot push " << value << "." << endl;
            return;
        }
        arr[++top] = value;
        cout << value << " pushed to stack." << endl;
    }

    // Pop an element from the stack
    int pop() {
        if (isEmpty()) {
            cout << "Error: Stack Underflow! Cannot pop from empty stack." << endl;
            return numeric_limits<int>::min(); // Return a sentinel value for error
        }
        int poppedValue = arr[top--];
        cout << poppedValue << " popped from stack." << endl;
        return poppedValue;
    }

    // Peek at the top element of the stack without removing it
    int peek() {
        if (isEmpty()) {
            cout << "Error: Stack is empty! Cannot peek." << endl;
            return numeric_limits<int>::min(); // Return a sentinel value for error
        }
        cout << "Top element is: " << arr[top] << endl;
        return arr[top];
    }

    // Display the elements of the stack
    void display() {
        if (isEmpty()) {
            cout << "Stack is empty." << endl;
            return;
        }
        cout << "Stack elements (top to bottom): ";
        for (int i = top; i >= 0; --i) {
            cout << arr[i] << " ";
        }
        cout << endl;
    }
};

int main() {
    Stack s;

    //cout << "\n--- Stack Operations ---\n";

    s.display();
    s.push(10);
    s.push(20);
    s.push(30);
    s.display();
    s.peek();
    s.pop();
    s.display();
    s.push(40);
    s.push(50);
    /*
    s.push(60);
    s.push(70);
    s.push(80);
    s.push(90);
    s.push(100);
    s.push(110); // Attempt to push when full
    */
    s.display();
    s.pop();
    s.pop();
    s.pop();
    /*
    s.display();
    s.pop();
    s.pop();
    s.pop();
    s.pop();
    s.pop();
    s.pop();
    */
    s.pop();
    s.pop(); // Attempt to pop when empty
    s.peek(); // Attempt to peek when empty

    return 0;
}
```
