# Experiment 1.
## Objective: Study and implementation of Breadth First Search (BFS) - Uninformed Search (Blind search)

## Theory
**Breadth First Search (BFS)** is an uninformed search algorithm that explores graph or tree structures level by level, visiting all nodes at the current depth before moving deeper.

* **Strategy**: Uses a First-In-First-Out (FIFO) queue for systematic, level-order exploration.
* **Completeness**: Guaranteed to find a solution in finite state spaces.
* **Optimality**: Guarantees the shortest path in unweighted graphs (all step costs equal).

#### Key Characteristics & Working
1. **FIFO Queue**: Maintains the search frontier; shallowest nodes are expanded first.
2. **Early Goal Test**: Evaluates goal conditions as soon as a node is generated to minimize unnecessary expansion.
3. **Redundancy Control**: Maintains an `explored` set to avoid infinite loops and duplicate state evaluation.

#### Pros & Cons
* **Advantages**: Simple implementation, guarantees shortest path on unweighted graphs, well-suited for shallow goals.
* **Limitations**: High memory requirements (stores all frontier nodes) and lacks sensitivity to edge weights.

#### Applications
* Web crawling, finding degrees of separation in social networks, network broadcasting, and State Space routing.

### Algorithm / Pseudocode

```
Function BREADTH-FIRST-SEARCH(problem) returns a solution node, or failure

    1. Create root node: node = node with STATE = problem.INITIAL-STATE, PATH-COST = 0
    2. if problem.GOAL-TEST(node.STATE) then return SOLUTION(node)
        
    3. Initialize frontier (FIFO queue) containing only node
    4. Initialize empty set: explored
        
    5. while not IS-EMPTY(frontier) do
        node = POP(frontier) /* removes shallowest node */
        add node.STATE to explored
            
        6. for each action in problem.ACTIONS(node.STATE) do
            child = CHILD-NODE(problem, node, action)
                
            7. if child.STATE is not in explored or frontier then
                if problem.GOAL-TEST(child.STATE) then return SOLUTION(child)
                INSERT(child, frontier)
                    
    return failure
```
### Program :

````python
# Python implementation of Breadth First Search (BFS)
from collections import deque

def bfs(graph, start, goal=None):
    """
    Performs a Breadth First Search on a graph from a start node to an optional goal node.
    """
    # 1 & 3. Initialize the frontier queue with the start node
    frontier = deque([start])

    # 4. Keep track of visited nodes to prevent redundant exploration
    explored = set([start])

    # Track parent nodes to reconstruct the path later
    parent = {start: None}

    visit_order = []

    if start == goal:
        return [start], visit_order

    while frontier:
        current_node = frontier.popleft()  # Dequeue the shallowest node (FIFO)
        visit_order.append(current_node)

        # 6. Check all neighbors/actions of the current node
        for neighbor in graph.get(current_node, []):
            # 7. If the neighbor hasn't been visited yet
            if neighbor not in explored:
                explored.add(neighbor)
                parent[neighbor] = current_node

                # Goal check as soon as the node is generated
                if neighbor == goal:
                    visit_order.append(neighbor)
                    # Reconstruct path
                    path = []
                    curr = goal
                    while curr is not None:
                        path.append(curr)
                        curr = parent[curr]
                    return path[::-1], visit_order

                frontier.append(neighbor)

    return None, visit_order

# --- Demonstration ---
# Defining an adjacency list representing a sample graph
sample_graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F', 'G'],
    'D': ['B'],
    'E': ['B', 'H'],
    'F': ['C'],
    'G': ['C'],
    'H': ['E']
}

start_node = 'A'
goal_node = 'H'

path, order = bfs(sample_graph, start_node, goal_node)

print(f"Start Node: {start_node}")
print(f"Goal Node: {goal_node}")
print(f"Nodes visited in order of exploration: {order}")
print(f"Shortest path found: {path}")
````

````python
%%writefile aione.cpp
#include <iostream>
#include <vector>
#include <queue>
#include <unordered_map>
#include <unordered_set>
#include <algorithm>

using namespace std;

// BFS implementation in C++
void bfs(const unordered_map<char, vector<char>>& graph, char start, char goal) {
    queue<char> frontier;
    unordered_set<char> explored;
    unordered_map<char, char> parent;
    vector<char> visit_order;

    frontier.push(start);
    explored.insert(start);
    parent[start] = '\0'; // Null/No parent indicator

    if (start == goal) {
        cout << "Shortest path found: " << start << endl;
        return;
    }

    bool goal_found = false;

    while (!frontier.empty()) {
        char current_node = frontier.front();
        frontier.pop();
        visit_order.push_back(current_node);

        for (char neighbor : graph.at(current_node)) {
            if (explored.find(neighbor) == explored.end()) {
                explored.insert(neighbor);
                parent[neighbor] = current_node;

                if (neighbor == goal) {
                    visit_order.push_back(neighbor);
                    goal_found = true;
                    break;
                }
                frontier.push(neighbor);
            }
        }
        if (goal_found) break;
    }

    cout << "Start Node: " << start << endl;
    cout << "Goal Node: " << goal << endl;
    cout << "Nodes visited in order of exploration: ";
    for (char node : visit_order) {
        cout << node << " ";
    }
    cout << endl;

    if (goal_found) {
        vector<char> path;
        char curr = goal;
        while (curr != '\0') {
            path.push_back(curr);
            curr = parent[curr];
        }
        reverse(path.begin(), path.end());

        cout << "Shortest path found: ";
        for (char node : path) {
            cout << node << " ";
        }
        cout << endl;
    } else {
        cout << "No path found." << endl;
    }
}

int main() {
    // Graph defined as an adjacency list
    unordered_map<char, vector<char>> sample_graph = {
        {'A', {'B', 'C'}},
        {'B', {'A', 'D', 'E'}},
        {'C', {'A', 'F', 'G'}},
        {'D', {'B'}},
        {'E', {'B', 'H'}},
        {'F', {'C'}},
        {'G', {'C'}},
        {'H', {'E'}}
    };

    bfs(sample_graph, 'A', 'H');
    return 0;
}
````

# Experiment 2.
## Objective: Study and implementation of Depth First Search (DFS) - Uninformed Search (Blind search)

## Theory
**Depth First Search (DFS)** is an uninformed search algorithm that explores a graph or tree structure by going as deep as possible along each branch before backtracking.

* **Strategy**: Uses a Last-In-First-Out (LIFO) stack (either the system call stack via recursion or an explicit stack data structure) to explore the deepest unvisited node.
* **Completeness**: Complete in finite state spaces. In infinite state spaces or graphs with loops, it can get stuck unless a cycle-detection mechanism (visited set) is used.
* **Optimality**: Not guaranteed to find the shallowest path or optimal solution first.

#### Key Characteristics & Working
1. **LIFO Stack**: The frontier is managed as a stack, prioritizing newly discovered deeper neighbors.
2. **Backtracking**: When a node has no unvisited neighbors left, the search backtracks to the most recently active ancestor node with remaining unvisited neighbors.
3. **Redundancy Control**: Like BFS, maintaining a `visited` or `explored` list is vital to handle cycles and prevent infinite loops.

#### Pros & Cons
* **Advantages**: Requires significantly less memory than BFS ($O(bm)$ space complexity, where $b$ is the branching factor and $m$ is maximum depth) because it only stores nodes on the current path.
* **Limitations**: Not optimal, and can get trapped in infinite paths if cycle checking or depth limits are not implemented.

#### Applications
* Solving puzzles (like mazes), topological sorting, cycle detection in graphs, and finding strongly connected components.

    Returns a solution node, or failure

    explored = an empty set
    
    while not IS-EMPTY(frontier) do
        node = POP(frontier) /* removes deepest node */
        
        if problem.GOAL-TEST(node.STATE) then return SOLUTION(node)

                child = CHILD-NODE(problem, node, action)
                if child.STATE is not in explored and not in frontier then
                    INSERT(child, frontier)
                    
    return failure

````python
# Python implementation of Depth First Search (DFS)

def dfs(graph, start, goal=None):
    """
    Performs a Depth First Search on a graph from a start node to an optional goal node.
    """
    # Use a list as a stack (LIFO) for frontier. Store tuple: (node, path_to_node)
    frontier = [(start, [start])]
    explored = set()
    visit_order = []

    while frontier:
        current_node, path = frontier.pop() # Pop from the end (LIFO)

        if current_node not in explored:
            visit_order.append(current_node)
            explored.add(current_node)

            # Goal check upon expansion
            if current_node == goal:
                return path, visit_order

            # Retrieve neighbors and add to stack
            # Reversing neighbors to match traditional left-to-right exploration order when popping from stack
            neighbors = graph.get(current_node, [])
            for neighbor in reversed(neighbors):
                if neighbor not in explored:
                    frontier.append((neighbor, path + [neighbor]))

    return None, visit_order

# --- Demonstration ---
sample_graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F', 'G'],
    'D': ['B'],
    'E': ['B', 'H'],
    'F': ['C'],
    'G': ['C'],
    'H': ['E']
}

start_node = 'A'
goal_node = 'H'

path, order = dfs(sample_graph, start_node, goal_node)

print(f"Start Node: {start_node}")
print(f"Goal Node: {goal_node}")
print(f"Nodes visited in order of exploration: {order}")
print(f"DFS Path found: {path}")
````

````python
%%writefile aitwo.cpp
#include <iostream>
#include <vector>
#include <stack>
#include <unordered_map>
#include <unordered_set>
#include <algorithm>

using namespace std;

// DFS implementation in C++ using an explicit stack
void dfs(const unordered_map<char, vector<char>>& graph, char start, char goal) {
    stack<pair<char, vector<char>>> frontier; // stores pair: {node, path_to_node}
    unordered_set<char> explored;
    vector<char> visit_order;

    frontier.push({start, {start}});
    bool goal_found = false;
    vector<char> final_path;

    while (!frontier.empty()) {
        auto current = frontier.top();
        frontier.pop();

        char current_node = current.first;
        vector<char> path = current.second;

        if (explored.find(current_node) == explored.end()) {
            visit_order.push_back(current_node);
            explored.insert(current_node);

            if (current_node == goal) {
                goal_found = true;
                final_path = path;
                break;
            }

            auto neighbors = graph.at(current_node);
            // Reverse neighbors to push onto the stack so we visit left-to-right
            for (auto it = neighbors.rbegin(); it != neighbors.rend(); ++it) {
                char neighbor = *it;
                if (explored.find(neighbor) == explored.end()) {
                    vector<char> next_path = path;
                    next_path.push_back(neighbor);
                    frontier.push({neighbor, next_path});
                }
            }
        }
    }

    cout << "Start Node: " << start << endl;
    cout << "Goal Node: " << goal << endl;
    cout << "Nodes visited in order of exploration: ";
    for (char node : visit_order) {
        cout << node << " ";
    }
    cout << endl;

    if (goal_found) {
        cout << "DFS Path found: ";
        for (char node : final_path) {
            cout << node << " ";
        }
        cout << endl;
    } else {
        cout << "No path found." << endl;
    }
}

int main() {
    unordered_map<char, vector<char>> sample_graph = {
        {'A', {'B', 'C'}},
        {'B', {'A', 'D', 'E'}},
        {'C', {'A', 'F', 'G'}},
        {'D', {'B'}},
        {'E', {'B', 'H'}},
        {'F', {'C'}},
        {'G', {'C'}},
        {'H', {'E'}}
    };

    dfs(sample_graph, 'A', 'H');
    return 0;
}
````
